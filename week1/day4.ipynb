```python
import pandas as pd

data = {
    "name": ["Alice", "Bob", "Cara", "Dan"],
    "age": [25, 30, 35, 28],
    "city": ["Stockholm", "Oslo", "Stockholm", "Berlin"]
}

df = pd.DataFrame(data)

print(df)
print(df.head())
print(df.describe())
print(df.info())
```

```python
print(df["age"])              # one column, as a pandas Series (like a labeled NumPy array)
print(df[["name", "age"]])    # multiple columns — note the double brackets

print(df.iloc[0])              # first row, by position
print(df.loc[0]) 
```

```python
older_than_28 = df[df["age"] > 28]
print(older_than_28)

stockholm_people = df[df["city"] == "Stockholm"]
print(stockholm_people)

# combine conditions with & (and) or | (or) — note the parentheses are required
result = df[(df["age"] > 25) & (df["city"] == "Stockholm")]
print(result)
```

```python
grouped = df.groupby("city")["age"].mean()
print(grouped)
```

```python
grouped_count = df.groupby("city")["age"].count()
print(grouped_count)
```

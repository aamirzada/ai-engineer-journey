```python
my_list = [1, 2, 4, 2]        # ordered, mutable, allows duplicates
my_tuple = (1, 4, 3)          # ordered, immutable
my_dict = {"a": 1, "b": 2, "c": 3}    # key-value pairs, mutable
my_set = {1, 2, 3, 2, 4, 2}         # unordered, mutable, no duplicates

print(my_list, my_tuple, my_dict, my_set)
```

```python
# the "long way"
squares = []
for n in range(10):
    squares.append(n ** 3)
print(squares)

# the comprehension way — same result, one line
squares = [n ** 2 for n in range(10)]
print(squares)
```

```python
even_squares = [n ** 2 for n in range(10) if n % 2 == 1]
print(even_squares)
```

```python
words = ["python", "ai", "engineer"]

word_lengths = {w: len(w) for w in words}        # dict comprehension
print(word_lengths)

unique_lengths = {len(w) for w in words}          # set comprehension
print(unique_lengths)
```

```python
evenodd = {n: {"even" if n % 2 == 0 else "odd"} for n in range(20)}  
print(evenodd)  # dict comprehension with condition
```

```python
import numpy as np

my_list = [1, 2, 3, 4]
my_array = np.array([1, 2, 3, 4])

print(my_list * 2)     # [1, 2, 3, 4, 1, 2, 3, 4]  — lists just repeat themselves
print(my_array * 2)   
```

```python
a = np.array([1, 2, 3])
zeros = np.zeros(5)              # [0. 0. 0. 0. 0.]
ones = np.ones((2, 3))           # a 2-row, 3-column grid of 1s
sequence = np.arange(0, 10, 2)   # [0 2 4 6 8] — like range(), but returns an array directly

print(a, zeros, ones, sequence, sep="\n")
```

```python
prices = np.array([10, 20, 30, 40])
discounted = prices * 0.9        # 10% off everything, no loop
print(discounted)                 # [ 9. 18. 27. 36.]

is_expensive = prices > 25
print(is_expensive)               # [False False  True  True]  — comparison on every element at once

print(prices[is_expensive])       # [30 40]  — use that True/False array to filter!
```

```python
grid = np.array([[1, 2, 3], [4, 5, 6]])
print(grid.shape)     # (2, 3) — 2 rows, 3 columns
print(grid[0])         # [1 2 3] — first row
print(grid[:, 1])      # [2 5]    — second column, every row
```

```python
arr = np.arange(25).reshape(5, 5)  # 5x5 grid of numbers from 0 to 24
print(arr)
arr = arr[::3, 0:3]  # slice the first 3 rows and first 3 columns
print(arr)
```

```python
arr = np.random.randint(0, 100, (5,5))
print(arr)

evens = arr[arr % 2 == 0]      # boolean mask → 1D array of even values
print(evens)
arr[:, -1] = 0 
print(arr)
```

```python
arr = [C * 9/5 + 32 for C in np.array([0, 10, 20, 30, 40])]
print(arr)
```
